```python
import sqlite3
from datetime import datetime

# Verbindung zur Datenbank
conn = sqlite3.connect("shop.db")
cursor = conn.cursor()

cursor.execute("PRAGMA foreign_keys = ON;")

# -------------------------
# Tabellen erstellen
# -------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS Lieferant (
    Lieferanten_ID INTEGER PRIMARY KEY,
    Name VARCHAR(20) NOT NULL
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Kunde (
    Kunden_ID INTEGER PRIMARY KEY,
    Name VARCHAR(20) NOT NULL,
    Adresse VARCHAR(20) NOT NULL
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Bestellung (
    Bestell_ID INTEGER PRIMARY KEY,
    Datum DATETIME NOT NULL,
    Kunden_ID INTEGER NOT NULL,
    FOREIGN KEY (Kunden_ID) REFERENCES Kunde(Kunden_ID)
        ON DELETE CASCADE ON UPDATE CASCADE
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Produkt (
    Produkt_ID INTEGER PRIMARY KEY,
    Name VARCHAR(20) NOT NULL,
    Kategorie VARCHAR(20),
    Preis DOUBLE NOT NULL,
    Lieferanten_ID INTEGER NOT NULL,
    Bestell_ID INTEGER NOT NULL,
    FOREIGN KEY (Lieferanten_ID) REFERENCES Lieferant(Lieferanten_ID)
        ON DELETE CASCADE ON UPDATE CASCADE,
    FOREIGN KEY (Bestell_ID) REFERENCES Bestellung(Bestell_ID)
        ON DELETE CASCADE ON UPDATE CASCADE
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Zahlung (
    Zahlungs_ID INTEGER PRIMARY KEY,
    Gesamtbetrag DOUBLE NOT NULL,
    Bestell_ID INTEGER UNIQUE NOT NULL,
    FOREIGN KEY (Bestell_ID) REFERENCES Bestellung(Bestell_ID)
        ON DELETE CASCADE ON UPDATE CASCADE
);
""")

# -------------------------
# Daten einzeln einfügen
# -------------------------

# Lieferanten
cursor.execute("INSERT INTO Lieferant VALUES (1, 'TechSupply')")
cursor.execute("INSERT INTO Lieferant VALUES (2, 'OfficePro')")
cursor.execute("INSERT INTO Lieferant VALUES (3, 'GadgetWorld')")

# Kunden
cursor.execute("INSERT INTO Kunde VALUES (1, 'Max Mustermann', 'Berlin')")
cursor.execute("INSERT INTO Kunde VALUES (2, 'Lisa Schmidt', 'Hamburg')")

# Bestellungen
cursor.execute("INSERT INTO Bestellung VALUES (1, ?, 1)", (datetime.now(),))
cursor.execute("INSERT INTO Bestellung VALUES (2, ?, 2)", (datetime.now(),))

# Produkte
cursor.execute("INSERT INTO Produkt VALUES (1, 'Laptop', 'Elektronik', 999.99, 1, 1)")
cursor.execute("INSERT INTO Produkt VALUES (2, 'Maus', 'Elektronik', 29.99, 1, 1)")
cursor.execute("INSERT INTO Produkt VALUES (3, 'Bürostuhl', 'Möbel', 199.99, 2, 2)")
cursor.execute("INSERT INTO Produkt VALUES (4, 'Monitor', 'Elektronik', 249.99, 3, 2)")

# Zahlungen
cursor.execute("INSERT INTO Zahlung VALUES (1, 1029.98, 1)")
cursor.execute("INSERT INTO Zahlung VALUES (2, 449.98, 2)")

conn.commit()

# -------------------------
# Testausgabe
# -------------------------

print("\nBestellübersicht:\n")
for row in cursor.execute("""
SELECT k.Name, b.Bestell_ID, p.Name, p.Preis, z.Gesamtbetrag
FROM Bestellung b
JOIN Kunde k ON b.Kunden_ID = k.Kunden_ID
JOIN Produkt p ON p.Bestell_ID = b.Bestell_ID
JOIN Zahlung z ON z.Bestell_ID = b.Bestell_ID
ORDER BY b.Bestell_ID;
"""):
    print(row)

conn.close()
```
